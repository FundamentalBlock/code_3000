# Discussion Questions

### Describe your reaction to the match rate. What sources of data might be prone to privacy issues?

The match rate shows that just three fields like age, zip code, and gender are enough to re-identify a large portion of records even without names. This is a problem because data like this gets shared or sold under the assumption that removing names makes it safe. Sources like voter registration records, hospital discharge data, and insurance records are especially at risk since they contain these same demographic fields and are often publicly accessible.


### Provide examples of at least two other variables that could increase the match rate if included.

Two variables that would increase the match rate are full date of birth and 5-digit ZIP code. Age in whole years groups many people together, so adding a birth month or full date would make most records unique on their own. The same applies to ZIP code, since a 3-digit prefix covers a large area and narrowing it to a full ZIP would make matches much more specific.
